| Data issue             | Example                     | Possible impact                      |
| ---------------------- | --------------------------- | ------------------------------------ |
| Missing `CustomerID`   | Customer not identified     | Customer analysis may be incomplete  |
| Negative quantity      | Returned or cancelled order | Revenue may be underestimated        |
| Duplicate rows         | Same transaction repeated   | Revenue may be counted twice         |
| Missing description    | Product name unavailable    | Product analysis becomes difficult   |
| Zero or negative price | Free product or data issue  | Revenue calculation may be incorrect |
